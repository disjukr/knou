### 문제 1

이름, 몸무게(kg), 키(cm)를 입력받아 BMI 수치와 상태를 반환하는 `BMI` 클래스를 완성하시오.

- BMI 수치 공식: **몸무게 ÷ (키(m) × 키(m))**  
  (키는 cm 단위로 입력되므로, 100으로 나누어 m 단위로 변환해야 합니다.)
- BMI 상태 기준:  
  - 18.5 미만 → `"저체중"`  
  - 18.5 이상 25.0 미만 → `"정상"`  
  - 25.0 이상 → `"과체중"`  

실행 결과:  
```
홍길동의 BMI: 24.51, 상태: 정상
김영희의 BMI: 17.59, 상태: 저체중
```

(힌트: 8강 객체지향 / Chapter 1 클래스의 정의, BMI 클래스)

**[답안 1]** 아래의 코드의 `?` 를 올바르게 수정하고, 실행하십시오.

````python
class BMI:
    def __init__(self, name, weight, height):
        self.name = name
        self.weight = weight
        self.height = height

    def get_BMI(self):
        return self.weight / (self.height / 100) ** 2

    def get_status(self):
        bmi = self.get_BMI()
        if bmi < 18.5:
            return "저체중"
        elif bmi < 25.0:
            return "정상"
        else:
            return "과체중"

# 테스트
p1 = BMI("홍길동", 70, 169)
p2 = BMI("김영희", 48, 165.2)

print(f"{p1.name}의 BMI: {p1.get_BMI():.2f}, 상태: {p1.get_status()}")
print(f"{p2.name}의 BMI: {p2.get_BMI():.2f}, 상태: {p2.get_status()}")
````

### 문제 3

PC 부품의 이름과 가격이 저장된 딕셔너리 `pc_parts`가 있습니다.  
`for` 반복문과 딕셔너리의 `items()` 메서드를 사용하여, 모든 부품의 가격 합계를 구하고 각 부품의 이름과 가격을 출력하는 코드를 완성하시오.

실행 결과:
```
CPU: 950,000원
RAM: 420,000원
SSD: 195,000원
GPU: 1,850,000원
--------------------
총 합계: 3,415,000원
```

(힌트: 10강 튜플, 세트, 딕셔너리 / Chapter 3 딕셔너리의 이해)

**[답안 3]** 아래의 코드의 `?` 를 올바르게 수정하고, 실행하십시오.

````python
pc_parts = {"CPU": 950000, "RAM": 420000, "SSD": 195000, "GPU": 1850000}

total = 0

for part, price in pc_parts.items():
    print(f"{part}: {price:,}원")
    total += price

print("-" * 20)
print(f"총 합계: {total:,}원")
````

### 문제 4

3명의 학생별 국어, 영어, 수학 점수가 2차원 리스트 `scores`에 저장되어 있습니다.  
중첩 `for` 반복문을 사용하여 각 학생의 점수 **합계**와 **평균**을 출력하는 코드를 완성하시오.

실행 결과:
```
홍길동: 합계=253, 평균=84.3
김영희: 합계=255, 평균=85.0
이민준: 합계=250, 평균=83.3
```

(힌트: 11강 다차원 구조 / Chapter 1 2차원 리스트의 이해, 2차원 리스트의 순회)

**[답안 4]** 아래의 코드의 `?` 를 올바르게 수정하고, 실행하십시오.

````python
# 3명의 학생별 국어, 영어, 수학 점수 (행: 학생, 열: 과목)
scores = [
    [85, 90, 78],   # 홍길동: 국어, 영어, 수학
    [72, 88, 95],   # 김영희: 국어, 영어, 수학
    [91, 76, 83],   # 이민준: 국어, 영어, 수학
]

names = ["홍길동", "김영희", "이민준"]

for i in range(len(scores)):              # 행(학생) 순회
    total = 0
    for j in range(len(scores[i])):
        total += scores[i][j]
    avg = total / len(scores[i])
    print(f"{names[i]}: 합계={total}, 평균={avg:.1f}")
````
